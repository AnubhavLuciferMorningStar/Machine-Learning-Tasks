```python
# merge
# Create the first DataFrame (e.g., Employee data)
df_employees = pd.DataFrame({
    'Emp_ID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Dept_ID': ['D1', 'D2', 'D1', 'D3', 'D5'] # Notice D5 doesn't exist in the department table
})

# Create the second DataFrame (e.g., Department data)
df_departments = pd.DataFrame({
    'Dept_ID': ['D1', 'D2', 'D3', 'D4'],      # Notice D4 has no employees
    'Department_Name': ['HR', 'Engineering', 'Finance', 'Marketing']
})

print("Employees DataFrame:")
display(df_employees)

print("\nDepartments DataFrame:")
display(df_departments)
```

```python
# inner merge
df_employees.merge(df_departments, on = 'Dept_ID', how = 'inner')
```

```python
# outer merge
df_employees.merge(df_departments, on = 'Dept_ID', how = 'outer')
```

```python
# left merge
df_employees.merge(df_departments, on = 'Dept_ID', how = 'left')
```

```python
# right merge
df_employees.merge(df_departments, on = 'Dept_ID', how = 'right')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

### Univariate

```python
# Histogram
sns.histplot(data = id_df, x = 'petal_length')
plt.show()
```

```python
# KDE Plot
sns.kdeplot(data = id_df, x = 'sepal_width')
plt.show()
```

```python
# Boxplot
sns.boxplot(data = id_df, x = 'sepal_length', color = 'skyblue')
plt.title('Boxplot of sepal length')
plt.xlabel('Sepal Length')
plt.show()
```

```python
# count plot
sns.countplot(data = id_df, x = "species")
plt.title('Countplot of Species')
plt.xlabel('Flower Species')
plt.show()
```

```python
# Calculate the proportions
species_counts = id_df['species'].value_counts()

# Create the pie chart
plt.pie(x=species_counts, 
        labels=species_counts.index, 
        autopct='%1.1f%%', 
        startangle=90, 
        colors=sns.color_palette('Set2'))

plt.title('Pie Chart of Species')
plt.show()
```

### Bivariate

```python
sns.lineplot(data = id_df, x = 'species', y = 'sepal_length')
plt.show()
```

```python
# 1. Create temporary time-series data
days = np.arange(1, 11)
product_a_sales = [20, 25, 22, 30, 28, 35, 40, 38, 45, 50]
product_b_sales = [15, 18, 25, 20, 30, 28, 32, 35, 40, 42]

# Combine into a Pandas DataFrame
temp_df = pd.DataFrame({
    'Day': np.tile(days, 2), # Repeats days 1-10 twice
    'Sales': product_a_sales + product_b_sales,
    'Product': ['Product A']*10 + ['Product B']*10
})

# 2. Create the lineplot with enhanced aesthetics
plt.figure(figsize=(8, 5))
sns.lineplot(
    data=temp_df, 
    x='Day', 
    y='Sales', 
    hue='Product', 
    marker='o',       # Adds circular dots to the actual data points
    linewidth=2.5,    # Makes the lines slightly thicker
    palette='Set2'    # Uses a nice pastel color palette
)

# 3. Add titles and tweak labels
plt.title('Product Sales Trend Over 10 Days', fontsize=14, pad=15)
plt.xlabel('Day', fontsize=12)
plt.ylabel('Sales Volume', fontsize=12)
plt.xticks(np.arange(1, 11, 1)) # Forces the X-axis to show exactly days 1 through 10

plt.show()
```
