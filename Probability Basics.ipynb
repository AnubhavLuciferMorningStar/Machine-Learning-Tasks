### What is Probability?
***Answer***: Probability measures the likelihood of an event occuring, expressed as a number 0(impossible) and 1(certain). Values between 0 and 1 represent partial chances.

***Formula of Probability:***
P(A) = Number of favourable outcomes to A/Total number of possible outcomes

* If P(A) = 0, then the event A is impossible.
* If P(A) = 1, then the event A is certain to occur.
* If 0<P(A)<1, then the event A is possible but not guranteed.

For example , When we toss a coin, there are only two possible outcomes: Heads (H) or Tails (T). However, if we toss two coins simultaneously, there will be four possible outcomes: (H, H), (H, T), (T, H), and (T, T).

***Experiment***: Experiment is any procedure that can be infinity repeated and has a well-defined set of possible outcomes, known as sample space(S).

***Event(E)***: Event is a subset of the sample space of an experiment, i.e., E <= S.

***Mutually Exclusive Events***: if two events are mutually exclusive, the the probability of both the events occuring at the same time is equal to zero, i.e.P(A intersect B) = 0.
***For example***: while tossing a coin, coming up heads and tails are the two mutually exclusive events.

***Mutually Exhaustive Events***: When tow events 'A' and 'B' exhaustive, it means that one of them must occur, i.e., p(A union B). ***For example***: while tossing a coin, coming up heads or tails are two mutually exclusive events.

***Independent Events***: Independent events means when the two events are independent if the occurance of one does not change  the probability of the other occuring. ***For Example***, If two events 'A' and 'B' are independent, then: P(A intersect B)  = P(A).P(B).

***Conditional Probability***: It is defined as the likelihood of an event or outcome occuring, based on the occurence of a previous event or outcome.

***For example***:
* P(event A will occur given that event B has already occured)
* P(A|B) = P(A union B)/P(B)
* Two events A and B are mutually independent if P(A|B).P(A). Also, P(A|B) != P(B|A)

### Basic rules of probability:
* Addition rule: P(A union B) = P(A) + P(B) - P(A intersect B)
* Multiplication Rule: P(A intersect B) = P(A|B).P(B) = P(B|A).P(A)
* Complement rule: P(A^c) = 1 - P(A)

***Bayes Theoram***: Conditional probability of each of a set of possible causes(B), given an observed outcome(A) is given as

### Types of a random variable
***Discreate Random Variable***: A randome variable with finite or countable number of possible values. e.g, random variable of the outcome when a dice is thrown. It can take integer value range 1 to 6.

***Continuous Random Variable***: A random variable that can take an infinite number of possible values. e.g, random variable representing the height of students in a class.

### Probability Distribution Function(PDF)
It is a function that is used to give the probability of all the possible values that a random variable can take. If the random variable is discrete, then it is called ***Probability Mass Function***. For the continuous variable, it is called ***Probability Density Function***.

### Cumulative Distribution Function(CDF)
It returns the probability that a random variable will take a value less than or equal to x.

### Expectation
Expectation of the discrete random variable(X) having a probability mass function P(X) is the weighted average possible valules that X can take. i,e. E(X) = n sigma 1 xi.P(xi)

***Note***: For a uniformaly distributed random variable. Expectation is equal to the mean.

***Properties of Expectation***:

E(aX) = a.E(X)

E(X+b) = E(X) + b

E(aX + b) = a.E(x) + b

### Python codes for implementing probability

```python
# Probability of getting an even number when rolling a dice

favorable_outcomes = 3
total_outcomes = 6

probability = favorable_outcomes / total_outcomes

print("Probability:", probability)
```

```python
# Probability of getting an even number

outcomes = [1, 2, 3, 4, 5, 6]

even_numbers = [num for num in outcomes if num % 2 == 0]

probability = len(even_numbers) / len(outcomes)

print("Even numbers:", even_numbers)
print("Probability:", probability)
```

```python
# Probability of getting Heads

total_outcomes = 2
favorable_outcomes = 1

probability_heads = favorable_outcomes / total_outcomes

print("Probability of Heads:", probability_heads)
```

```python
# Simulating Coin Tosses
import random
number_of_tosses = 1000
results = []
for i in range(number_of_tosses):
    result = random.choice(["Heads", "Tails"])
    results.append(result)
heads_count = results.count("Heads")
tails_count = results.count("Tails")

probability_of_heads = heads_count/number_of_tosses
probability_of_tails = tails_count/number_of_tosses

print("Heads:", heads_count)
print("Tails:", tails_count)

print("Probability of Heads:", probability_of_heads)
print("Probability of Tails:", probability_of_tails)
```

```python
# Dice Roll Simulation
number_of_rolls = 100
dice_results = []
for i in range(number_of_rolls):
    dice = random.randint(1,6)
    dice_results.append(dice)

# probability of getting 6
count_six = dice_results.count(6)
probability_six = count_six

print("Number of times 6 occured:", count_six)
print("Probability of getting 6:", probability_six)

# P(6)=1/6≈0.1667
```

```python
# Complement Rule
# The complement rule is: P(A^c)=1−P(A)
# Example: Probability of not getting 6 on a dice.
probability_getting_six = 1 / 6

probability_not_getting_six = 1 - probability_getting_six

print("Probability of getting 6:", probability_getting_six)
print("Probability of not getting 6:", probability_not_getting_six)
```

```python
# Addition Rule - A or B
# P(A union B) = P(A) + P(B)
probability_1 = 1/6
probability_2 = 1/6
probability_of_getiing_1_or_2 = probability_1 + probability_2
print("Probability of getting 1 or 2:", probability_of_getiing_1_or_2)
```

```python
# General Addition Rule
# P(A union B) = P(A) + P(B) + P(A intersect B)
P_A = 10
P_B = 7
P_A_N_B = 20
P_A_U_B = P_A + P_B + P_A_N_B
print("Probability of A union B:", P_A_U_B)
```

```python
# Multiplication Rule
# P(A intersect B) = P(A) * P(B)
# Probability of geeting heads twice in a row
probability_heads = 1 / 2
probability_two_heads = probability_heads * probability_heads
print("Probability of getting Heads twice:", probability_two_heads)
```

```python
#Conditional Probability
# Formula: P(A|B) = P(A intersect B)/P(B)
P_A_N_B = 30
P_B = 20
P_A_given_B = P_A_N_B / P_B
print("P(A|B):", P_A_given_B)
```

```python
# Checking Whether Two Events Are Independent
# Two events are independent if: P(A intersect B) = P(A) * P(B)
P_A = 0.72
P_B = 0.36
P_A_N_B = 0.18
expected_probability = P_A * P_B
if P_A_N_B == expected_probability:
    print("Events are independent")
else:
    print("Events are dependent")
```

```python
# Probability Using NumPy
import numpy as np

# Simulate 10,000 coin tosses
coin_tosses = np.random.choice(
    ["Heads", "Tails"],
    size=10000
)

probability_heads = np.mean(coin_tosses == "Heads")
probability_tails = np.mean(coin_tosses == "Tails")

print("Probability of Heads:", probability_heads)
print("Probability of Tails:", probability_tails)
```
