```python
# 4.Count Digits
n = int(input())

count = 0
temp = abs(n)

# Handle the edge case where the input is exactly 0
if temp == 0:
    count = 1
else:
    # Divide by 10 and count each step
    while temp > 0:
        count += 1
        temp //= 10

print(count)
```

```python
# 5. Palindrome Number
# Check whether a number is palindrome.
# Example:
# 121 → Palindrome
# 123 → Not Palindrome
n = int(input("Enter a number: "))

# Convert to string and check if it equals its reverse
if str(n) == str(n)[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
# 6. Armstrong Number
# Check whether a number is Armstrong.
# Example:
# 153 → Armstrong
n = int(input())

# Convert to string to easily get the number of digits (power) and iterate
num_str = str(n)
power = len(num_str)

# Sum each digit raised to the power
armstrong_sum = sum(int(digit) ** power for digit in num_str)

if armstrong_sum == n:
    print("Armstrong")
else:
    print("Not Armstrong")
```

```python
# 8. Fibonacci Series
# Print first N Fibonacci numbers.
# Example:
# 0 1 1 2 3 5 8
def fibonacci(n):
    a, b = 0, 1
    result = []
    for _ in range(n):
        result.append(a)
        a, b = b, a + b
    return result

n = int(input("Enter N: "))
print(*fibonacci(n))
```

```python
# 9. Prime Number Checker
# Check if a number is prime using function.
def check_prime(n):
    if n<=0:
        return "Not Prime"
    for i in range(2, int(n**0.5)+1):
        if n %1==0:
            return "Not Prime"
    return "Prime"

n = int(input())
print(check_prime(n))
```

```python
# 11. Greatest Common Divisor (GCD)
# Find GCD of two numbers.
# Example:
# 12, 18 → 6
def find_gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return abs(a)

a = 12
b = 18
print(find_gcd(a, b))
```

```python
# 12. Least Common Multiple (LCM)
# Find LCM of two numbers.

def find_lcm(a, b):
    # The LCM of any number and 0 is 0
    if a == 0 or b == 0:
        return 0
    a, b = abs(a), abs(b)
    larger = max(a, b)
    smaller = min(a, b)

    multiple = larger
    while multiple % smaller != 0:
        multiple += larger

    return multiple

a = 18
b = 15
print(find_lcm(a,b))
```

```python
# 13. Perfect Number
# Check whether a number is perfect.
# Example:
# 6 → 1+2+3 = 6

def is_perfect(n):
    # Perfect numbers are strictly positive integers
    if n <= 0:
        return False
        
    divisor_sum = 0
    # We only need to check up to n // 2, as no divisor can be larger than that
    for i in range(1, (n // 2) + 1):
        if n % i == 0:
            divisor_sum += i
            
    return divisor_sum == n

# Example usage
n = int(input())

if is_perfect(n):
    print(f"{n} is a Perfect Number")
else:
    print(f"{n} is Not a Perfect Number")
```

```python
# 14. Strong Number
# Check whether a number is strong.
#Example: 145 → 1! + 4! + 5! = 145

def is_strong_number(n):
    # Function to calculate factorial without imports
    def get_factorial(digit):
        fact = 1
        for i in range(1, digit + 1):
            fact *= i
        return fact

    # Sum the factorials of all digits
    digit_factorial_sum = sum(get_factorial(int(digit)) for digit in str(n))
    return digit_factorial_sum == n

# Example usage
n = int(input())
if is_strong_number(n):
    print(f"{n} is a Strong Number.")
else:
    print(f"{n} is NOT a Strong Number.")
```

```python
# 15. Pattern Program
# Print:
#    1
#   12
#  123
# 1234
#12345
n = int(input())
for i in range(1, n+1):
    for j in range(n-i):
        print(" ", end = "")
    for j in range(1, i+1):
        print(j, end = "")
    print()
```

```python
# 18. Count Vowels
# Count vowels in a string.
# Input: "programming"
# Output: 3
def count_vowels(A):
        v = 0
        for i in A:
            if i=="a" or i=="e" or i=="i" or i=="o" or i=="u":
                v+=1
        return v
    
A = "Programming"
print(count_vowels(A))
```

```python
# 19. Remove Duplicates from List
# Input: [1,2,2,3,4,4,5]
# Output: [1,2,3,4,5]
def remove_duplicates(lst):
    result = []
    for item in lst:
        if item not in result:
            result.append(item)
    return result

print(remove_duplicates([1, 2, 2, 3, 4, 4, 5]))
```

```python
# 20. Calculator Using Functions
# Create functions for:
# add()
# subtract()
# multiply()
# divide()
# Menu-driven calculator.
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Error: Division by zero is not allowed"
    return a / b


def calculator():
    while True:
        print("\n--- Calculator Menu ---")
        print("1. Add")
        print("2. Subtract")
        print("3. Multiply")
        print("4. Divide")
        print("5. Exit")

        choice = input("Enter your choice (1-5): ")

        if choice == '5':
            print("Exiting calculator. Goodbye!")
            break

        if choice not in ('1', '2', '3', '4'):
            print("Invalid choice. Please try again.")
            continue

        try:
            num1 = float(input("Enter first number: "))
            num2 = float(input("Enter second number: "))
        except ValueError:
            print("Invalid input. Please enter numbers only.")
            continue

        if choice == '1':
            print(f"Result: {add(num1, num2)}")
        elif choice == '2':
            print(f"Result: {subtract(num1, num2)}")
        elif choice == '3':
            print(f"Result: {multiply(num1, num2)}")
        elif choice == '4':
            print(f"Result: {divide(num1, num2)}")


if __name__ == "__main__":
    calculator()
```
